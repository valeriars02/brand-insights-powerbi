## Research question

How did average brand familiarity and liking change from 2020 to 2024 for Netflix, Humana, Pfizer, M·A·C, and Boston Scientific?

## Data source

This project uses the BRAND (Brand Recognition and Attitude Norms Database), available through [ResearchBox](https://researchbox.org/1892). The dataset is licensed under CC BY 4.0.

Original study: Raffaelli et al., “BRAND: Brand recognition and attitude norms database,” *Behavior Research Methods*. [https://doi.org/10.3758/s13428-024-02525-x](https://doi.org/10.3758/s13428-024-02525-x)

## Scope and limitations

The data contains aggregated responses from U.S. residents. This notebook compares average familiarity and liking scores for five selected brands. The comparisons are descriptive: they do not establish why a score changed or show that the same people responded in both years.

## 2. Inspect the workbook and worksheet

The workbook structure is checked before loading the data. This helps identify which worksheet contains the dataset. The workbook has one worksheet, named `BRAND`.

```python
#Nombres de las hojas del Excel
import pandas as pd
archivo = "BRAND_dataset.xlsx"
excel = pd.ExcelFile(archivo)
excel.sheet_names
```

```python
#Filas y coulmnas que contiene el dataset
df=pd.read_excel(archivo, sheet_name="BRAND")
print("Filas y columnas:", df.shape)
print("\nNombre de las columnas:")
print(df.columns.tolist())

display(df.head())
```

## 3. Inspect the table layout and headers

The worksheet uses three header rows: the data section, the measure, and the statistic or field name. Setting `header=None` prevents pandas from treating only the first row as the column names. The transposed preview makes it easier to inspect the headers and the first brand record. This step does not modify the data.

```python
raw = pd.read_excel(archivo, sheet_name="BRAND", header=None)
display(raw.iloc[:4,:].T)
```

## 4. Display the headers in a readable format

The three header rows are displayed as a table with one row per original column. Forward filling carries section and measure labels across grouped columns, making the structure easier to understand. This creates a readable preview; it does not change the source data.

```python
#Encabezados
encabezados = raw.iloc[:3,:].T.copy()
encabezados.columns =["Sección","Medida","Campo"]

encabezados["Sección"]=encabezados["Sección"].ffill()
encabezados["Medida"]=encabezados["Medida"].ffill()

display(encabezados)
```

## 5. Prepare the analysis table

The first three rows contain the column headers, so they are combined into descriptive column names. The brand records begin after those header rows. The preparation removes only columns that are completely empty and rows without a brand name; partial missing values are kept for review.

The resulting table contains 597 brand records and 56 columns.

```python
# Completar hacia la derecha los encabezados agrupados
nivel_1 = raw.iloc[0].ffill()
nivel_2 = raw.iloc[1].ffill()
nivel_3 = raw.iloc[2]

# Crear nombres de columnas combinando los tres niveles
nombres = []
for i in range(raw.shape[1]):
    partes = [
        str(x).strip()
        for x in [nivel_1.iloc[i], nivel_2.iloc[i], nivel_3.iloc[i]]
        if pd.notna(x) and str(x).strip() != ""
    ]
    nombres.append(" | ".join(partes))

# Los datos empiezan después de las tres filas de encabezados
data_limpia = raw.iloc[3:].copy().reset_index(drop=True)
data_limpia.columns = nombres

# Quitar solo columnas completamente vacías y filas sin nombre de marca
data_limpia = data_limpia.dropna(axis=1, how="all")
data_limpia = data_limpia.dropna(subset=["Brand Name"])

print("Dimensiones:", data_limpia.shape)

faltantes = pd.DataFrame({
    "faltantes": data_limpia.isna().sum(),
    "porcentaje": (data_limpia.isna().mean() * 100).round(1)
})

display(
    faltantes[faltantes["faltantes"] > 0]
    .sort_values("porcentaje", ascending=False)
)

display(data_limpia.head(3).iloc[:, :12])
```

## 6. Investigate missing data by year

This step identifies brands with no survey measures recorded for 2020 or 2024. These missing values appear to reflect which brands were included in each year's study, so they are not automatically errors. They should not be imputed or removed without considering the analysis.

```python
columnas_2020 = [c for c in data_limpia.columns if c.startswith("BRAND 2020 |")]
columnas_2024 = [c for c in data_limpia.columns if c.startswith("BRAND 2024 |")]

sin_2020 = data_limpia[columnas_2020].isna().all(axis=1)
sin_2024 = data_limpia[columnas_2024].isna().all(axis=1)

print("Marcas sin medidas de 2020:", sin_2020.sum())
display(data_limpia.loc[sin_2020, ["Brand Name", "Category"]].head(15))

print("Marcas sin medidas de 2024:", sin_2024.sum())
display(data_limpia.loc[sin_2024, ["Brand Name", "Category"]])
```

## 7. Select brands measured in both years

To compare 2020 and 2024, this step creates a separate table containing only brands with a familiarity score in both years. It includes 492 brands. The full `data_limpia` table remains unchanged, and Netflix is displayed as one example.

```python
familiaridad_2020 = "BRAND 2020 | Brand Familiarity | M"
familiaridad_2024 = "BRAND 2024 | Brand Familiarity | M"

marcas_comparables = data_limpia.dropna(
    subset=[familiaridad_2020, familiaridad_2024]
).copy()

print("Marcas con familiaridad en ambos años:", len(marcas_comparables))

display(
    data_limpia.loc[
        data_limpia["Brand Name"].eq("Netflix"),
        ["Brand Name", "Category", familiaridad_2020, familiaridad_2024]
    ]
)
```

## 8. Compare familiarity changes across brands

For each brand measured in both years, familiarity change is calculated as the 2024 mean minus the 2020 mean. Positive values indicate a higher average familiarity score in 2024; negative values indicate a lower score. The tables show the largest increases and decreases. These are descriptive comparisons and do not explain why the scores changed.

```python
marcas_comparables["Cambio_familiaridad"] = (
    marcas_comparables[familiaridad_2024]
    - marcas_comparables[familiaridad_2020]
)

columnas_vista = [
    "Brand Name",
    "Category",
    familiaridad_2020,
    familiaridad_2024,
    "Cambio_familiaridad"
]

print("Mayores aumentos:")
display(
    marcas_comparables[columnas_vista]
    .sort_values("Cambio_familiaridad", ascending=False)
    .head(10)
)

print("Mayores disminuciones:")
display(
    marcas_comparables[columnas_vista]
    .sort_values("Cambio_familiaridad")
    .head(10)
)
```

## 9. Inspect response counts and variability

This step displays the number of responses (`n`), average familiarity (`M`), and response variability (`SD`) for candidate brands in both years. These statistics add context to the changes in averages: they show how many responses contributed to each score and how spread out those responses were.

```python
marcas_revisar = [
    "Boston Scientific", "Pfizer", "Humana",
    "Netflix", "M·A·C", "Progressive"
]

columnas = [
    "Brand Name",
    "Category",
    "BRAND 2020 | Brand Familiarity | n",
    "BRAND 2020 | Brand Familiarity | M",
    "BRAND 2020 | Brand Familiarity | SD",
    "BRAND 2024 | Brand Familiarity | n",
    "BRAND 2024 | Brand Familiarity | M",
    "BRAND 2024 | Brand Familiarity | SD",
    "Cambio_familiaridad"
]

display(
    marcas_comparables.loc[
        marcas_comparables["Brand Name"].isin(marcas_revisar),
        columnas
    ]
)
```

## 10. Compare brand liking across the selected brands

This table compares average brand liking for Netflix, Humana, Pfizer, M·A·C, and Boston Scientific in 2020 and 2024. It includes the number of responses (`n`), average score (`M`), and response variability (`SD`) for each year.

```python
marcas_seleccionadas = [
    "Netflix",
    "Humana",
    "Pfizer",
    "M·A·C",
    "Boston Scientific"
]

columnas_gusto = [
    "Brand Name",
    "Category",
    "BRAND 2020 | Brand Liking | n",
    "BRAND 2020 | Brand Liking | M",
    "BRAND 2020 | Brand Liking | SD",
    "BRAND 2024 | Brand Liking | n",
    "BRAND 2024 | Brand Liking | M",
    "BRAND 2024 | Brand Liking | SD"
]

tabla_gusto = data_limpia.loc[
    data_limpia["Brand Name"].isin(marcas_seleccionadas),
    columnas_gusto
]

gusto_2020 = "BRAND 2020 | Brand Liking | M"
gusto_2024 = "BRAND 2024 | Brand Liking | M"

tabla_gusto = tabla_gusto.copy()
tabla_gusto["Cambio_gusto"] = (
    tabla_gusto[gusto_2024] - tabla_gusto[gusto_2020]
).round(3)

display(tabla_gusto)
```

## 11. Prepare and export the Power BI dataset

This step selects the five brands and the familiarity and liking measures for 2020 and 2024. It calculates the change in each average score and exports the results as a XLSX file for Power BI.

```python
marcas_seleccionadas = [
    "Netflix", "Humana", "Pfizer", "M·A·C", "Boston Scientific"
]

columnas = [
    "Brand Name",
    "Category",
    "BRAND 2020 | Brand Familiarity | n",
    "BRAND 2020 | Brand Familiarity | M",
    "BRAND 2020 | Brand Familiarity | SD",
    "BRAND 2024 | Brand Familiarity | n",
    "BRAND 2024 | Brand Familiarity | M",
    "BRAND 2024 | Brand Familiarity | SD",
    "BRAND 2020 | Brand Liking | n",
    "BRAND 2020 | Brand Liking | M",
    "BRAND 2020 | Brand Liking | SD",
    "BRAND 2024 | Brand Liking | n",
    "BRAND 2024 | Brand Liking | M",
    "BRAND 2024 | Brand Liking | SD"
]

powerbi_data = data_limpia.loc[
    data_limpia["Brand Name"].isin(marcas_seleccionadas),
    columnas
].copy()

powerbi_data = powerbi_data.rename(columns={
    "Brand Name": "Brand",
    "BRAND 2020 | Brand Familiarity | n": "Familiarity_n_2020",
    "BRAND 2020 | Brand Familiarity | M": "Familiarity_mean_2020",
    "BRAND 2020 | Brand Familiarity | SD": "Familiarity_SD_2020",
    "BRAND 2024 | Brand Familiarity | n": "Familiarity_n_2024",
    "BRAND 2024 | Brand Familiarity | M": "Familiarity_mean_2024",
    "BRAND 2024 | Brand Familiarity | SD": "Familiarity_SD_2024",
    "BRAND 2020 | Brand Liking | n": "Liking_n_2020",
    "BRAND 2020 | Brand Liking | M": "Liking_mean_2020",
    "BRAND 2020 | Brand Liking | SD": "Liking_SD_2020",
    "BRAND 2024 | Brand Liking | n": "Liking_n_2024",
    "BRAND 2024 | Brand Liking | M": "Liking_mean_2024",
    "BRAND 2024 | Brand Liking | SD": "Liking_SD_2024"
})

powerbi_data["Familiarity_change"] = (
    powerbi_data["Familiarity_mean_2024"]
    - powerbi_data["Familiarity_mean_2020"]
)

powerbi_data["Liking_change"] = (
    powerbi_data["Liking_mean_2024"]
    - powerbi_data["Liking_mean_2020"]
)

display(powerbi_data)

powerbi_data.to_excel(
    "brand_insights_powerbi.xlsx",
    index=False
)

from google.colab import files
files.download("brand_insights_powerbi.xlsx")
```

## 12. Summary and limitations

### Summary

Among the five selected brands, average familiarity increased the most for Boston Scientific (+1.88 points), followed by Pfizer (+1.73) and Humana (+1.43). M·A·C decreased (-1.43), while Netflix remained nearly unchanged (+0.01).

Average liking decreased for Netflix (-0.49), Pfizer (-0.28), and Humana (-0.18). M·A·C remained unchanged, while Boston Scientific increased (+0.54).

These are descriptive differences between average scores in 2020 and 2024. They do not establish why the scores changed.

### Limitations

- The survey responses came from U.S. residents, so the results should not be generalized to consumers in Mexico.
- This analysis includes five selected brands; it does not represent their entire industries.
- The number of responses varies by brand, measure, and year.
- The scores are aggregated by year and do not track the same individuals over time.
